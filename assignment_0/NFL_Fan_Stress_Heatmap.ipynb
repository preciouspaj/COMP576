```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib
import matplotlib as mpl

teams = ["Cowboys", "Eagles", "Chiefs", "Bills", "Texans"]

times = [
    "Pregame",
    "1st Quarter",
    "Halftime",
    "4th Quarter",
    "Final 2 Mins"
]

stress_data = np.array([
    [100, 100, 100, 100, 100],
    [42, 50, 61, 83, 90],
    [30, 38, 47, 70, 79],
    [48, 60, 72, 89, 95],
    [40, 52, 65, 80, 87]
])

fig, ax = plt.subplots()

im = ax.imshow(stress_data)

# Show all ticks and label them with the respective list entries
ax.set_xticks(
    range(len(times)),
    labels=times,
    rotation=45,
    rotation_mode="default"
)

ax.set_yticks(
    range(len(teams)),
    labels=teams
)

# Loop over data dimensions and create text annotations
for i in range(len(teams)):
    for j in range(len(times)):
        text = ax.text(
            j,
            i,
            stress_data[i, j],
            ha="center",
            va="center",
            color="w"
        )

ax.set_title("NFL Fan Stress Level")

fig.tight_layout()

plt.show()
```
